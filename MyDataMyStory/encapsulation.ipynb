```python
class Book:
    def __init__(self, book_id, title, author):
        self.book_id = book_id
        self.title = title
        self.author = author
        self.is_available = True  # True means available, False means borrowed


class Library:
    def __init__(self):
        self.books = []
        # Tracks how many books each student currently has borrowed
        self.student_borrow_counts = {}

    def add_book(self, book):
        self.books.append(book)
        print(f"Added: '{book.title}' by {book.author}")

    def issue_book(self, book_id, student_id):
        # Challenge check: Maximum 3 books per student
        current_borrowed = self.student_borrow_counts.get(student_id, 0)
        if current_borrowed >= 3:
            print(f"Error: Student {student_id} has reached the maximum limit of 3 borrowed books.")
            return

        for book in self.books:
            if book.book_id == book_id:
                if not book.is_available:
                    print(f"Error: '{book.title}' is already borrowed.")
                    return
                # Issue the book
                book.is_available = False
                self.student_borrow_counts[student_id] = current_borrowed + 1
                print(f"Success: '{book.title}' issued to student {student_id}.")
                return
        print("Error: Book ID not found.")

    def return_book(self, book_id, student_id):
        for book in self.books:
            if book.book_id == book_id:
                if book.is_available:
                    print(f"Error: '{book.title}' was not marked as borrowed.")
                    return
                # Return the book
                book.is_available = True
                self.student_borrow_counts[student_id] = max(0, self.student_borrow_counts.get(student_id, 1) - 1)
                print(f"Success: '{book.title}' returned by student {student_id}.")
                return
        print("Error: Book ID not found.")

    def search_by_title(self, title):
        results = [b for b in self.books if title.lower() in b.title.lower()]
        if results:
            print(f"\nSearch results for '{title}':")
            for b in results:
                status = "Available" if b.is_available else "Borrowed"
                print(f"- [{b.book_id}] {b.title} by {b.author} ({status})")
        else:
            print(f"No books found matching '{title}'.")

    def display_available_books(self):
        print("\n--- Available Books ---")
        available_books = [b for b in self.books if b.is_available]
        if not available_books:
            print("No books currently available.")
        for b in available_books:
            print(f"[{b.book_id}] {b.title} by {b.author}")


# --- Example Usage ---
lib = Library()
lib.add_book(Book("B101", "Python Programming", "Guido van Rossum"))
lib.add_book(Book("B102", "Data Structures", "Alice Smith"))
lib.add_book(Book("B103", "Algorithms", "Bob Jones"))
lib.add_book(Book("B104", "Database Systems", "Charlie Brown"))

# Issue books to student 'S101'
lib.issue_book("B101", "S101")
lib.issue_book("B102", "S101")
lib.issue_book("B103", "S101")

# Attempting to issue a 4th book triggers the challenge restriction
lib.issue_book("B104", "S101") 

lib.display_available_books()
```

```python
class Book:
    def __init__(self, book_id, title, author):
        self.book_id = book_id
        self.title = title
        self.author = author
        self.available = True


class Library:
    def __init__(self):
        self.books = []

    def add_book(self, book):
        self.books.append(book)
        print("Book added successfully!")

    def issue_book(self, title):
        for book in self.books:
            if book.title.lower() == title.lower():
                if book.available:
                    book.available = False
                    print("Book issued successfully!")
                else:
                    print("Book is already borrowed!")
                return
        print("Book not found!")

    def return_book(self, title):
        for book in self.books:
            if book.title.lower() == title.lower():
                if not book.available:
                    book.available = True
                    print("Book returned successfully!")
                else:
                    print("Book was not issued!")
                return
        print("Book not found!")

    def search_book(self, title):
        for book in self.books:
            if book.title.lower() == title.lower():
                print("Book ID:", book.book_id)
                print("Title:", book.title)
                print("Author:", book.author)
                print("Availability:", "Available" if book.available else "Borrowed")
                return
        print("Book not found!")

    def display_books(self):
        print("\nAvailable Books:")
        for book in self.books:
            if book.available:
                print(book.book_id, book.title, book.author)


# Main Program
library = Library()

library.add_book(Book(1, "Python", "Guido"))
library.add_book(Book(2, "Java", "James Gosling"))
library.add_book(Book(3, "C++", "Bjarne Stroustrup"))

library.display_books()

library.issue_book("Python")

library.search_book("Python")

library.return_book("Python")

library.display_books()
```

```python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

class Cube(Shape):
    def __init__(self, width, height, length):
        self.width = width
        self.height = height
        self.length = length

    def area(self):
        return self.width * self.height * self.length

c = Cube(2,10, 4)
print(c.area())
print(c.area())
```
